```python
# write a function to load data
import json

def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)
        return data
```

```python
data = load_data("data.json")
```

```python
data
```

```python
# function to display user and their connection 
def display_user(data):
    for user in data['users']: 
        print(f"{user['id']}. {user['name']} is friend with: {user['friends']} and liked pages are {user['liked_pages']}")
    print("pages information")
    for page in data['pages']:
        print(f"{page['id']}: {page['name']}")

display_user(data)
```
